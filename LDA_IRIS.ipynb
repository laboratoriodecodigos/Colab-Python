```python
# Análisis Discriminante Canónico (LDA/CDA)

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# Cargar dataset Iris
iris = load_iris()

X = iris.data
y = iris.target

# Crear modelo CDA
cda = LinearDiscriminantAnalysis(n_components=2)

# Transformar datos
X_cda = cda.fit_transform(X, y)

# Graficar resultados
plt.figure(figsize=(8,5))

for clase, nombre in enumerate(iris.target_names):
    plt.scatter(
        X_cda[y == clase, 0],
        X_cda[y == clase, 1],
        label=nombre
    )

plt.xlabel("Función Canónica 1")
plt.ylabel("Función Canónica 2")
plt.title("Análisis Discriminante Canónico")
plt.legend()
plt.grid(True)
plt.show()

# Importancia de las variables
coef = pd.DataFrame(
    cda.coef_,
    columns=iris.feature_names
)

print("Coeficientes discriminantes:")
print(coef)
```
