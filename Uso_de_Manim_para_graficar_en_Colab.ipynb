```python
%%manim -qm -v WARNING RegresionML
# Calidad media para un renderizado rápido

from manim import *
import numpy as np

class RegresionML(Scene):
  def construct(self):
      # 1. Configuración de Ejes
      ejes = Axes(
          x_range=[0, 6, 1],
          y_range=[0, 6, 1],
          x_length=7,
          y_length=5,
          axis_config={"color": WHITE}
      )

      # Etiquetas usando Text
      etiqueta_x = Text("Característica (X)", font_size=24).next_to(ejes.x_axis, DOWN)
      etiqueta_y = Text("Predicción (Y)", font_size=24).rotate(90 * DEGREES).next_to(ejes.y_axis, LEFT)

      # 2. Generar puntos de datos "aleatorios"
      puntos_coord = [
          [1, 1.5], [2, 1.8], [3, 3.2], [4, 3.8], [5, 4.5], [1.5, 2.5], [3.5, 2.8]
      ]

      puntos = VGroup(*[
          Dot(ejes.c2p(x, y), color=BLUE) for x, y in puntos_coord
      ])

      # 3. La línea de regresión (Modelo de ML)
      # Iniciamos con una línea plana y luego la "entrenamos"
      linea_modelo = ejes.plot(lambda x: 0.5 * x + 0.5, color=YELLOW)
      texto_modelo = Text("Modelo de Regresión", color=YELLOW, font_size=30).to_corner(DOWN + RIGHT)

      # --- Secuencia de Animación ---

      # Presentación
      self.play(Write(Text("Machine Learning: Regresión", font_size=40).to_edge(UP)))
      self.play(Create(ejes), Write(etiqueta_x), Write(etiqueta_y))
      self.wait(0.5)

      # Aparecen los datos
      self.play(Create(puntos), run_time=2, lag_ratio=0.1)
      self.wait(1)

      # Aparece el modelo intentando ajustar
      self.play(Write(texto_modelo))
      self.play(Create(linea_modelo), run_time=2)

      # Simulación de "Ajuste/Entrenamiento"
      # (Cambiamos la pendiente de la línea para que se vea como optimización)
      linea_mejorada = ejes.plot(lambda x: 0.8 * x + 0.2, color=GREEN)
      self.play(
          Transform(linea_modelo, linea_mejorada),
          texto_modelo.animate.set_color(GREEN),
          run_time=2
      )

      self.wait(2)
```

```python
%%manim -qm -v WARNING GraficaLaboratorio
# -qm: Calidad Media (720p).
# -v WARNING: Limpia la consola de mensajes innecesarios.

from manim import *

class GraficaLaboratorio(Scene):
    def construct(self):
        # 1. Configuración de los ejes (Números estándar, no requieren LaTeX)
        ejes = Axes(
            x_range=[0, 10, 1],
            y_range=[-1.5, 1.5, 1],
            x_length=10,
            y_length=6,
            axis_config={"color": WHITE},
            tips=True
        )

        # 2. Definimos la función matemática
        grafica = ejes.plot(lambda x: np.sin(x), color=BLUE)

        # 3. Etiquetas usando 'Text' en lugar de 'MathTex'
        # Usamos texto plano para representar la fórmula
        titulo = Text("Laboratorio de Códigos", font_size=36).to_edge(UP)

        # Cambiamos MathTex por Text
        formula = Text("f(x) = sin(x)", font_size=28).next_to(titulo, DOWN, buff=0.2).set_color(YELLOW)

        # --- Secuencia de Animación ---

        # Animación del título
        self.play(AddTextLetterByLetter(titulo))

        # Creación de los ejes
        self.play(Create(ejes), run_time=2)
        self.wait(0.5)

        # Animación de la curva y el texto de la fórmula simultáneamente
        self.play(
            Create(grafica),
            Write(formula),
            run_time=3
        )

        self.wait(2)
```
