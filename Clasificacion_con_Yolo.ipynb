```python
import cv2
import urllib.request
from ultralytics import YOLO

# Cargar el modelo YOLOv8 preentrenado (entrenado en el dataset COCO con 80 clases)
model = YOLO('yolov8n.pt')
```

```python
print(model.names)
```

```python
videos_demo = {
    "personas_y_vehiculos": "https://github.com/intel-iot-devkit/sample-videos/raw/master/person-bicycle-car-detection.mp4",
    "personas_y_vehiculos_2": "videoplayback,mp4",
    "animales": "videoperros.mp4"

}
```

```python
import urllib.parse

# Descargar los videos en el entorno local de Colab
for nombre, url in videos_demo.items():
    # Check if the URL is a valid web URL before attempting to download
    parsed_url = urllib.parse.urlparse(url)
    if parsed_url.scheme in ['http', 'https', 'ftp']:
        archivo_destino = f"{nombre}.mp4"
        print(f"Descargando {nombre} desde {url}...")
        try:
            urllib.request.urlretrieve(url, archivo_destino)
            print(f"✅ {nombre} descargado correctamente.")
        except Exception as e:
            print(f"❌ Error descargando {nombre} desde {url}: {e}")
    else:
        print(f"⏩ Saltando {nombre}: '{url}' no es una URL de descarga válida. Podría ser un archivo local existente o una URL mal formada.")

print("¡Videos descargados con éxito!")
```

```python
def procesar_video(video_entrada, video_salida):

    cap = cv2.VideoCapture(video_entrada)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = int(cap.get(cv2.CAP_PROP_FPS))

    # Configurar el grabador de video
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(video_salida, fourcc, fps, (width, height))

    frame_count = 0
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        results = model(frame, verbose=False)

        annotated_frame = results[0].plot()

        out.write(annotated_frame)
        frame_count += 1

    cap.release()
    out.release()

# Procesar los dos videos descargados
procesar_video("personas_y_vehiculos.mp4", "resultado_personas_vehiculos.mp4")
procesar_video("animales.mp4", "resultado_animales.mp4")
```
