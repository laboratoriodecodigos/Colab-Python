```python
from time import sleep

# Sudoku de ejemplo (0 = espacio vacío)
sudoku = [
   [7, 8, 0, 4, 0, 0, 1, 2, 0],
    [6, 0, 0, 0, 7, 5, 0, 0, 9],
    [0, 0, 0, 6, 0, 1, 0, 7, 8],
    [0, 0, 7, 0, 4, 0, 2, 6, 0],
    [0, 0, 1, 0, 5, 0, 9, 3, 0],
    [9, 0, 4, 0, 6, 0, 0, 0, 5],
    [0, 7, 0, 3, 0, 0, 0, 1, 2],
    [1, 2, 0, 0, 0, 7, 4, 0, 0],
    [0, 4, 9, 2, 0, 6, 0, 0, 7]
]
```

```python
# =========================================
# IMPRIMIR SUDOKU
# =========================================
def imprimir(tablero):
    print("\n" + "=" * 25)

    for fila in range(9):

        if fila % 3 == 0 and fila != 0:
            print("-" * 25)

        for col in range(9):

            if col % 3 == 0 and col != 0:
                print("| ", end="")

            print(tablero[fila][col], end=" ")

        print()

    print("=" * 25)
```

```python
imprimir(sudoku)
```

```python
# =========================================
# BUSCAR ESPACIO VACÍO
# =========================================
def buscar_vacio(tablero):

    for fila in range(9):
        for col in range(9):

            if tablero[fila][col] == 0:
                return fila, col

    return None
```

```python
buscar_vacio(sudoku)
```

```python
# =========================================
# VALIDAR MOVIMIENTO
# =========================================
def es_valido(tablero, num, posicion):

    fila, col = posicion

    # Revisar fila
    for i in range(9):
        if tablero[fila][i] == num and col != i:
            return False

    # Revisar columna
    for i in range(9):
        if tablero[i][col] == num and fila != i:
            return False

    # Revisar cuadrante 3x3
    cuadro_x = col // 3
    cuadro_y = fila // 3

    for i in range(cuadro_y * 3, cuadro_y * 3 + 3):
        for j in range(cuadro_x * 3, cuadro_x * 3 + 3):

            if tablero[i][j] == num and (i, j) != posicion:
                return False

    return True
```

```python
# =========================================
# ALGORITMO BACKTRACKING
# =========================================
def resolver(tablero):

    encontrado = buscar_vacio(tablero)

    # Caso base: Sudoku completo
    if not encontrado:
        return True

    fila, col = encontrado

    # Probar números del 1 al 9
    for num in range(1, 10):

        if es_valido(tablero, num, (fila, col)):

            tablero[fila][col] = num

            # Llamada recursiva
            if resolver(tablero):
                return True

            # BACKTRACKING:
            # si no funciona, regresamos
            tablero[fila][col] = 0

    return False
```

```python
# =========================================
# PROGRAMA PRINCIPAL
# =========================================

print("SUDOKU ORIGINAL")
imprimir(sudoku)

if resolver(sudoku):

    print("\nSUDOKU RESUELTO")
    imprimir(sudoku)

else:
    print("No existe solución.")
```

```python
%%manim -ql ExplicacionBacktracking

from manim import *

class ExplicacionBacktracking(MovingCameraScene):
    def construct(self):
        # 1. TÍTULO E INTRODUCCIÓN
        titulo = Text("Algoritmo de Backtracking", font_size=40, color=BLUE)
        subtitulo = Text("Exploración con Retorno", font_size=24, color=GRAY)
        subtitulo.next_to(titulo, DOWN)

        self.play(Write(titulo), FadeIn(subtitulo, shift=DOWN))
        self.wait(2)
        self.play(FadeOut(titulo), FadeOut(subtitulo))

        # 2. CONFIGURACIÓN DEL ÁRBOL DE DECISIÓN
        # Definimos posiciones manuales para un árbol de decisiones de 3 niveles
        posiciones = {
            "Raiz": ORIGIN + UP * 2.5,
            "A": LEFT * 2.5 + UP * 0.5,
            "B": RIGHT * 2.5 + UP * 0.5,
            "A1": LEFT * 3.5 + DOWN * 1.5,
            "A2": LEFT * 1.5 + DOWN * 1.5,
            "B1": RIGHT * 1.5 + DOWN * 1.5,
            "B2": RIGHT * 3.5 + DOWN * 1.5,
        }

        # Crear los nodos (Círculos)
        nodos = {k: Circle(radius=0.4, color=WHITE, fill_opacity=1, fill_color=BLACK).move_to(pos)
                 for k, pos in posiciones.items()}

        # Etiquetas de texto dentro de los nodos
        etiquetas = {
            "Raiz": Text("Inicio", font_size=16).move_to(posiciones["Raiz"]),
            "A": Text("Opción A", font_size=14).move_to(posiciones["A"]),
            "B": Text("Opción B", font_size=14).move_to(posiciones["B"]),
            "A1": Text("Error", font_size=14, color=RED).move_to(posiciones["A1"]),
            "A2": Text("Éxito", font_size=14, color=GREEN).move_to(posiciones["A2"]),
            "B1": Text("-", font_size=14).move_to(posiciones["B1"]),
            "B2": Text("-", font_size=14).move_to(posiciones["B2"]),
        }

        # Crear las líneas (Ramas del árbol)
        ramas = [
            Line(posiciones["Raiz"], posiciones["A"], stroke_width=3),
            Line(posiciones["Raiz"], posiciones["B"], stroke_width=3),
            Line(posiciones["A"], posiciones["A1"], stroke_width=3),
            Line(posiciones["A"], posiciones["A2"], stroke_width=3),
            Line(posiciones["B"], posiciones["B1"], stroke_width=3),
            Line(posiciones["B"], posiciones["B2"], stroke_width=3),
        ]

        # Mostrar estructura inicial del Árbol de Decisión
        self.play(FadeIn(nodos["Raiz"]), Write(etiquetas["Raiz"]))
        self.play(
            *[Create(r) for r in ramas],
            *[FadeIn(nodos[k]) for k in nodos if k != "Raiz"],
            *[Write(etiquetas[k]) for k in etiquetas if k != "Raiz"]
        )
        self.wait(1)

        # 3. ANIMACIÓN DEL ALGORITMO (PASO A PASO)

        # Enfocar la cámara al inicio
        self.camera.frame.save_state()
        self.play(self.camera.frame.animate.set(width=10).move_to(posiciones["A"]))

        # Paso 1: Avanzar a la Opción A (Válida temporalmente)
        self.play(nodos["Raiz"].animate.set_color(GREEN), run_time=0.5)
        self.play(nodos["A"].animate.set_color(GREEN), run_time=0.8)
        self.wait(0.5)

        # Paso 2: Avanzar a A1 (Callejón sin salida / Error)
        self.play(self.camera.frame.animate.move_to(posiciones["A1"]))
        self.play(nodos["A1"].animate.set_color(RED).set_fill(RED, opacity=0.3), run_time=0.8)

        # Cartel de Alerta de Restricción Rota
        alerta = Text("¡Rompe las reglas!", font_size=18, color=RED).next_to(nodos["A1"], DOWN)
        self.play(Write(alerta))
        self.wait(1.5)
        self.play(FadeOut(alerta))

        # Paso 3: ¡BACKTRACK! (Retroceder a la Opción A)
        flecha_backtrack = Arrow(posiciones["A1"], posiciones["A"], color=ORANGE, stroke_width=5)
        txt_backtrack = Text("¡Backtrack!", font_size=20, color=ORANGE).next_to(flecha_backtrack, RIGHT)

        self.play(self.camera.frame.animate.move_to(posiciones["A"]))
        self.play(GrowArrow(flecha_backtrack), Write(txt_backtrack))
        self.wait(1)
        # Limpiamos el nodo erróneo (el algoritmo borra la decisión)
        self.play(FadeOut(flecha_backtrack), FadeOut(txt_backtrack))
        self.play(nodos["A1"].animate.set_color(GRAY).set_fill(BLACK), run_time=0.5)

        # Paso 4: Probar el siguiente camino (A2)
        self.play(self.camera.frame.animate.move_to(posiciones["A2"]))
        self.play(nodos["A2"].animate.set_color(GREEN).set_fill(GREEN, opacity=0.3), run_time=0.8)

        # Cartel de Solución Encontrada
        exito_txt = Text("Solución Válida", font_size=20, color=GREEN).next_to(nodos["A2"], DOWN)
        self.play(Write(exito_txt))
        self.wait(1)

        # 4. RESTAURAR CÁMARA Y CIERRE
        self.play(Restore(self.camera.frame))
        self.wait(2)
```

```python
%%manim -ql BacktrackingCompleto

from manim import *

class BacktrackingCompleto(MovingCameraScene):
    def construct(self):
        # ==========================================
        # ESCENA 1: EL LABERINTO (Duración: 18 segundos)
        # ==========================================
        # Creamos una analogía visual de un laberinto simple usando líneas horizontales y verticales
        laberinto = VGroup(
            Line(LEFT*3 + UP*2, RIGHT*3 + UP*2),
            Line(LEFT*3 + UP*2, LEFT*3 + DOWN*2),
            Line(RIGHT*3 + UP*2, RIGHT*3 + DOWN*2),
            Line(LEFT*1 + UP*2, LEFT*1 + DOWN*0.5),
            Line(RIGHT*1 + UP*2, RIGHT*1 + DOWN*0.5),
            Line(LEFT*3 + DOWN*2, LEFT*1 + DOWN*2),
            Line(RIGHT*1 + DOWN*2, RIGHT*3 + DOWN*2),
        ).set_color(GRAY_D)

        # Bifurcación y caminos
        bifurcacion_dot = Dot(UP*1.2, color=BLUE, radius=0.15)
        bifurcacion_txt = Text("Bifurcación", font_size=16, color=BLUE).next_to(bifurcacion_dot, UP*0.5)

        # Puntos de trayectoria
        camino_incorrecto = Line(UP*1.2, LEFT*2 + UP*1.2, color=RED, stroke_width=6)
        pared_bloqueo = Cross(scale_factor=0.3).move_to(LEFT*2 + UP*1.2).set_color(RED)

        camino_retorno = Line(LEFT*2 + UP*1.2, UP*1.2, color=ORANGE, stroke_width=4)
        camino_correcto = Line(UP*1.2, DOWN*1.5, color=GREEN, stroke_width=6)

        # Animación Escena 1
        self.play(Create(laberinto), run_time=3)
        self.play(FadeIn(bifurcacion_dot), Write(bifurcacion_txt), run_time=2)

        # Avanza y choca con pared
        self.play(Create(camino_incorrecto), run_time=3)
        self.play(Create(pared_bloqueo), run_time=1)
        self.wait(2) # Pausa dramática: "¿Qué haces? No te quedas ahí..."

        # Regresa y toma el otro camino
        self.play(Create(camino_retorno), run_time=2.5)
        self.play(Create(camino_correcto), run_time=3.5)
        self.wait(1) # Total acumulado del bloque: ~18 segundos

        # Limpieza para la siguiente escena
        self.play(FadeOut(laberinto), FadeOut(bifurcacion_dot), FadeOut(bifurcacion_txt),
                  FadeOut(camino_incorrecto), FadeOut(pared_bloqueo), FadeOut(camino_retorno), FadeOut(camino_correcto))

        # ==========================================
        # ESCENA 2: DEFINICIÓN INFORMÁTICA (Duración: 13 segundos)
        # ==========================================
        titulo = Text("Algoritmo de Backtracking", font_size=36, color=BLUE)
        def_1 = Text("Método sistemático de búsqueda", font_size=22, color=WHITE).next_to(titulo, DOWN, buff=0.5)
        def_2 = Text("Prueba opciones hasta encontrar la correcta", font_size=22, color=WHITE).next_to(def_1, DOWN)

        sub_caja = VGroup(titulo, def_1, def_2).move_to(ORIGIN)

        self.play(Write(titulo), run_time=3)
        self.play(FadeIn(def_1, shift=UP), run_time=2.5)
        self.play(FadeIn(def_2, shift=UP), run_time=2.5)
        self.wait(5) # Mantenemos el texto en pantalla para completar la explicación. Total: 13 segundos

        self.play(FadeOut(sub_caja))

        # ==========================================
        # ESCENA 3: EL ÁRBOL DE DECISIÓN (Duración: 13 segundos)
        # ==========================================
        # Definición geométrica de un árbol de expansión
        pos_raiz = UP * 2.5
        pos_izq = LEFT * 2 + UP * 0.5
        pos_der = RIGHT * 2 + UP * 0.5

        nodo_raiz = Circle(radius=0.4, color=BLUE, fill_opacity=1, fill_color=BLACK).move_to(pos_raiz)
        lbl_raiz = Text("Raíz (Inicio)", font_size=14).move_to(pos_raiz)

        nodo_izq = Circle(radius=0.4, color=WHITE).move_to(pos_izq)
        lbl_izq = Text("Opción 1", font_size=12).move_to(pos_izq)

        nodo_der = Circle(radius=0.4, color=WHITE).move_to(pos_der)
        lbl_der = Text("Opción 2", font_size=12).move_to(pos_der)

        rama_izq = Line(pos_raiz, pos_izq)
        rama_der = Line(pos_raiz, pos_der)

        # Animación Escena 3
        self.play(FadeIn(nodo_raiz), Write(lbl_raiz), run_time=3)
        self.wait(2) # Explicando la raíz

        # Se abren las ramas y opciones
        self.play(
            Create(rama_izq), Create(rama_der),
            FadeIn(nodo_izq), FadeIn(nodo_der),
            Write(lbl_izq), Write(lbl_der),
            run_time=4
        )
        self.wait(4) # Total: 13 segundos

        # ==========================================
        # ESCENA 4: REGLAS Y RESTRICCIONES (Duración: 16 segundos)
        # ==========================================
        # Expandimos el árbol hacia abajo (Sub-ramas de la Opción 1)
        pos_sub1 = LEFT * 3 + DOWN * 1.5
        pos_sub2 = LEFT * 1 + DOWN * 1.5

        nodo_sub1 = Circle(radius=0.4, color=WHITE).move_to(pos_sub1)
        lbl_sub1 = Text("Fallo", font_size=12).move_to(pos_sub1)

        nodo_sub2 = Circle(radius=0.4, color=WHITE).move_to(pos_sub2)
        lbl_sub2 = Text("Éxito", font_size=12).move_to(pos_sub2)

        rama_sub1 = Line(pos_izq, pos_sub1)
        rama_sub2 = Line(pos_izq, pos_sub2)

        # Mover la cámara dinámicamente para simular la exploración hacia abajo
        self.camera.frame.save_state()
        self.play(self.camera.frame.animate.set(width=8).move_to(LEFT*1.5 + DOWN*0.5), run_time=3)

        # Avanza con confianza por la Opción 1
        self.play(nodo_izq.animate.set_color(GREEN), run_time=2)
        self.play(Create(rama_sub1), FadeIn(nodo_sub1), Write(lbl_sub1), run_time=2)

        # Encuentra que rompe las reglas (Se vuelve rojo y se tacha)
        self.play(
            nodo_sub1.animate.set_color(RED).set_fill(RED, opacity=0.3),
            lbl_sub1.animate.set_color(RED),
            run_time=2
        )

        # Una gran "X" para indicar la cancelación de la rama completa
        tacha_rama = Cross(stroke_width=6, scale_factor=0.6).move_to(pos_sub1)
        txt_restriccion = Text("¡Restricción Rota!", font_size=14, color=RED).next_to(nodo_sub1, DOWN*0.5)

        self.play(Create(tacha_rama), Write(txt_restriccion), run_time=2)
        self.wait(5) # Mantener escena final para ajustar la locución exacta. Total: 16 segundos

        # Regresar cámara al estado original
        self.play(Restore(self.camera.frame), run_time=2)
        self.wait(1)
```

```python
%%manim -ql BacktrackingCompleto

from manim import *

class BacktrackingCompleto(MovingCameraScene):
    def construct(self):
        # ==========================================
        # ESCENA 1: EL LABERINTO (Duración: 18 segundos)
        # ==========================================
        laberinto = VGroup(
            Line(LEFT*3 + UP*2, RIGHT*3 + UP*2),
            Line(LEFT*3 + UP*2, LEFT*3 + DOWN*2),
            Line(RIGHT*3 + UP*2, RIGHT*3 + DOWN*2),
            Line(LEFT*1 + UP*2, LEFT*1 + DOWN*0.5),
            Line(RIGHT*1 + UP*2, RIGHT*1 + DOWN*0.5),
            Line(LEFT*3 + DOWN*2, LEFT*1 + DOWN*2),
            Line(RIGHT*1 + DOWN*2, RIGHT*3 + DOWN*2),
        ).set_color(GRAY_D)

        bifurcacion_dot = Dot(UP*1.2, color=BLUE, radius=0.15)
        bifurcacion_txt = Text("Bifurcación", font_size=16, color=BLUE).next_to(bifurcacion_dot, UP*0.5)

        camino_incorrecto = Line(UP*1.2, LEFT*2 + UP*1.2, color=RED, stroke_width=6)
        pared_bloqueo = Cross(scale_factor=0.3).move_to(LEFT*2 + UP*1.2).set_color(RED)

        camino_retorno = Line(LEFT*2 + UP*1.2, UP*1.2, color=ORANGE, stroke_width=4)
        camino_correcto = Line(UP*1.2, DOWN*1.5, color=GREEN, stroke_width=6)

        self.play(Create(laberinto), run_time=3)
        self.play(FadeIn(bifurcacion_dot), Write(bifurcacion_txt), run_time=2)

        self.play(Create(camino_incorrecto), run_time=3)
        self.play(Create(pared_bloqueo), run_time=1)
        self.wait(2)

        self.play(Create(camino_retorno), run_time=2.5)
        self.play(Create(camino_correcto), run_time=3.5)
        self.wait(1)

        self.play(FadeOut(laberinto), FadeOut(bifurcacion_dot), FadeOut(bifurcacion_txt),
                  FadeOut(camino_incorrecto), FadeOut(pared_bloqueo), FadeOut(camino_retorno), FadeOut(camino_correcto))

        # ==========================================
        # ESCENA 2: DEFINICIÓN INFORMÁTICA (Duración: 13 segundos)
        # ==========================================
        titulo = Text("Algoritmo de Backtracking", font_size=36, color=BLUE)
        def_1 = Text("Método sistemático de búsqueda", font_size=22, color=WHITE).next_to(titulo, DOWN, buff=0.5)
        def_2 = Text("Prueba opciones hasta encontrar la correcta", font_size=22, color=WHITE).next_to(def_1, DOWN)

        sub_caja = VGroup(titulo, def_1, def_2).move_to(ORIGIN)

        self.play(Write(titulo), run_time=3)
        self.play(FadeIn(def_1, shift=UP), run_time=2.5)
        self.play(FadeIn(def_2, shift=UP), run_time=2.5)
        self.wait(5)

        self.play(FadeOut(sub_caja))

        # ==========================================
        # ESCENA 3: EL ÁRBOL DE DECISIÓN (Duración: 13 segundos)
        # ==========================================
        pos_raiz = UP * 2.5
        pos_izq = LEFT * 2 + UP * 0.5
        pos_der = RIGHT * 2 + UP * 0.5

        nodo_raiz = Circle(radius=0.4, color=BLUE, fill_opacity=1, fill_color=BLACK).move_to(pos_raiz)
        lbl_raiz = Text("Raíz (Inicio)", font_size=14).move_to(pos_raiz)

        nodo_izq = Circle(radius=0.4, color=WHITE).move_to(pos_izq)
        lbl_izq = Text("Opción 1", font_size=12).move_to(pos_izq)

        nodo_der = Circle(radius=0.4, color=WHITE).move_to(pos_der)
        lbl_der = Text("Opción 2", font_size=12).move_to(pos_der)

        rama_izq = Line(pos_raiz, pos_izq)
        rama_der = Line(pos_raiz, pos_der)

        self.play(FadeIn(nodo_raiz), Write(lbl_raiz), run_time=3)
        self.wait(2)

        self.play(
            Create(rama_izq), Create(rama_der),
            FadeIn(nodo_izq), FadeIn(nodo_der),
            Write(lbl_izq), Write(lbl_der),
            run_time=4
        )
        self.wait(4)

        # ==========================================
        # ESCENA 4: REGLAS Y RESTRICCIONES (Duración: 16 segundos)
        # ==========================================
        pos_sub1 = LEFT * 3 + DOWN * 1.5
        pos_sub2 = LEFT * 1 + DOWN * 1.5

        nodo_sub1 = Circle(radius=0.4, color=WHITE).move_to(pos_sub1)
        lbl_sub1 = Text("Fallo", font_size=12).move_to(pos_sub1)

        nodo_sub2 = Circle(radius=0.4, color=WHITE).move_to(pos_sub2)
        lbl_sub2 = Text("Éxito", font_size=12).move_to(pos_sub2)

        rama_sub1 = Line(pos_izq, pos_sub1)
        rama_sub2 = Line(pos_izq, pos_sub2)

        self.camera.frame.save_state()
        self.play(self.camera.frame.animate.set(width=8).move_to(LEFT*1.5 + DOWN*0.5), run_time=3)

        self.play(nodo_izq.animate.set_color(GREEN), run_time=2)
        self.play(Create(rama_sub1), FadeIn(nodo_sub1), Write(lbl_sub1), run_time=2)

        self.play(
            nodo_sub1.animate.set_color(RED).set_fill(RED, opacity=0.3),
            lbl_sub1.animate.set_color(RED),
            run_time=2
        )

        tacha_rama = Cross(stroke_width=6, scale_factor=0.6).move_to(pos_sub1)
        txt_restriccion = Text("¡Restricción Rota!", font_size=14, color=RED).next_to(nodo_sub1, DOWN*0.5)

        self.play(Create(tacha_rama), Write(txt_restriccion), run_time=2)
        self.wait(5)

        # ==========================================
        # ESCENA 5: EL RETORNO / BACKTRACK (Nueva - Duración: ~15 segundos)
        # ==========================================
        # Eliminamos el texto de restricción anterior
        self.play(FadeOut(txt_restriccion), run_time=1)

        # 1. "Aquí ocurre la magia: el algoritmo hace Backtrack"
        # Flecha curva o recta indicando el retroceso físico hacia el nodo padre (Opción 1)
        flecha_retroceso = Arrow(pos_sub1, pos_izq, color=ORANGE, stroke_width=5, buff=0.1)
        txt_backtrack = Text("¡Backtrack!", font_size=16, color=ORANGE).next_to(flecha_retroceso, RIGHT, buff=0.1)

        self.play(GrowArrow(flecha_retroceso), Write(txt_backtrack), run_time=2)
        self.wait(1)

        # 2. "Vuelve un paso atrás en el árbol, borra la última mala decisión..."
        # Desvanecemos el rastro de la mala decisión y movemos el foco de la cámara de nuevo arriba
        self.play(
            FadeOut(flecha_retroceso), FadeOut(txt_backtrack),
            nodo_sub1.animate.set_color(GRAY_C).set_fill(BLACK, opacity=0),
            lbl_sub1.animate.set_color(GRAY_C),
            FadeOut(tacha_rama),
            self.camera.frame.animate.move_to(LEFT*1.5 + DOWN*0.5), # Mantiene centrado el subárbol
            run_time=3
        )

        # 3. "...y prueba la siguiente alternativa."
        # Creamos y recorremos la rama hacia el nodo de Éxito
        self.play(Create(rama_sub2), FadeIn(nodo_sub2), Write(lbl_sub2), run_time=2)
        self.play(
            nodo_sub2.animate.set_color(GREEN).set_fill(GREEN, opacity=0.4),
            lbl_sub2.animate.set_color(GREEN),
            run_time=2
        )

        # 4. "Así, evita explorar caminos inútiles de forma inteligente."
        txt_inteligente = Text("Búsqueda Optimizada", font_size=14, color=GREEN).next_to(nodo_sub2, DOWN*0.5)
        self.play(Write(txt_inteligente), run_time=1)
        self.wait(3) # Pausa final para terminar la frase con fuerza

        # Restaurar la cámara al plano completo del árbol al finalizar todo
        self.play(Restore(self.camera.frame), run_time=2)
        self.wait(1)
```
