```python
import os
from google.colab import userdata
from scrapegraphai.graphs import SmartScraperGraph
import json
```

```python
# 1. Configurar la API Key de OpenAI
try:
    os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
except Exception:
    import getpass
    os.environ["OPENAI_API_KEY"] = getpass.getpass("Ingresa tu OpenAI API Key: ")

# 2. Configurar el motor de IA
graph_config = {
    "llm": {
        "model": "openai/gpt-4o-mini",
        "temperature": 0,
    },
    "headless": True,
    "playwright_params": {
        "args": ["--no-sandbox", "--disable-dev-shm-usage"]
    }
}

# 3. URL de la página web a analizar
target_url = "https://books.toscrape.com/"

# 4. Prompt en lenguaje natural
prompt = """
Extrae una lista con los primeros 5 libros que aparecen en la página.
Para cada libro incluye:
- Título del libro.
- Precio (con su moneda).
- Calificación o rating (estrellas o texto).
- Estado de disponibilidad (si está en stock o no).

Devuelve el resultado en un arreglo de objetos JSON.
"""

# 5. Ejecutar el scraping
print("⏳ Leyendo la página web e interpretando el contenido...")
smart_scraper = SmartScraperGraph(
    prompt=prompt,
    source=target_url,
    config=graph_config
)

result = smart_scraper.run()

# 6. Imprimir resultado
print("\n✅ Información extraída:")
print(json.dumps(result, indent=2, ensure_ascii=False))
```
