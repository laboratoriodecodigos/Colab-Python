```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
dataframe=pd.read_excel('dataframe1.xlsx')
```

```python
display(dataframe)
```

```python
dataframe.info()
```

```python
dataframe.isnull().sum()
```

```python
num_cols = [
    "edad", "imc", "colesterol_total", "hdl_colesterol",
    "ldl_colesterol", "trigliceridos", "glucosa_ayunas",
    "hemoglobina_glicada"
]

cat_cols = ["genero", "presion_arterial"]
```

```python
imputer_num = SimpleImputer(strategy="median")
imputer_cat = SimpleImputer(strategy="most_frequent")
```

```python
dataframe[num_cols] = imputer_num.fit_transform(dataframe[num_cols])
```

```python
dataframe[cat_cols] = imputer_cat.fit_transform(dataframe[cat_cols])
```

```python
dataframe.isnull().sum()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot numerical columns
plt.figure(figsize=(15, 12))
plt.suptitle('Distributions of Numerical Variables', fontsize=16)
for i, col in enumerate(num_cols):
    plt.subplot(3, 3, i + 1)  # 3 rows, 3 columns
    sns.histplot(dataframe[col], kde=True)
    plt.title(f'{col}')
    plt.xlabel('') # Remove x-label to avoid clutter
plt.tight_layout(rect=[0, 0.03, 1, 0.96]) # Adjust layout to prevent suptitle overlap
plt.show()

# Plot categorical columns
plt.figure(figsize=(12, 5))
plt.suptitle('Counts of Categorical Variables', fontsize=16)
for i, col in enumerate(cat_cols):
    plt.subplot(1, 2, i + 1)
    sns.countplot(x=dataframe[col])
    plt.title(f'{col}')
    plt.xlabel('') # Remove x-label to avoid clutter
plt.tight_layout(rect=[0, 0.03, 1, 0.92]) # Adjust layout to prevent suptitle overlap
plt.show()
```
