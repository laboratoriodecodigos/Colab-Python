```python
from transformers import MarianMTModel, MarianTokenizer

# Modelo de traducción Español -> Inglés
model_name = "Helsinki-NLP/opus-mt-es-en"

tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)
```

```python
def traducir(texto):
    tokens = tokenizer(texto, return_tensors="pt", padding=True)
    traduccion = model.generate(**tokens)
    resultado = tokenizer.decode(traduccion[0], skip_special_tokens=True)
    return resultado
```

```python
texto = "La inteligencia artificial está transformando la educación y la industria."

traduccion = traducir(texto)

print("Texto original:", texto)
print("Traducción:", traduccion)
```

```python
textos = [
    "La ciencia de datos es una de las profesiones del futuro.",
    "Los transformers revolucionaron el procesamiento de lenguaje natural.",
    "Google Colab permite ejecutar modelos de inteligencia artificial fácilmente."
]

for t in textos:
    print("Original:", t)
    print("Traducción:", traducir(t))
    print()
```
