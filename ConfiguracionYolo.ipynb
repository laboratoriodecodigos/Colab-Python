```python
import os
from google.colab import files
from ultralytics import YOLO

# 1. Usamos el modelo Medium (descarga automática la primera vez)
model = YOLO("yolo11m.pt")

url_video = "https://raw.githubusercontent.com/intel-iot-devkit/sample-videos/master/person-bicycle-car-detection.mp4"

# 2. Inferencia mejorada
results = model.predict(
    source=url_video,
    save=True,
    conf=0.25,         # Captura objetos más lejanos o parcialmente ocluidos
    iou=0.45,          # Elimina cajas duplicadas sobre el mismo auto
    classes=[0, 2],    # 0: Personas, 2: Autos
    augment=True       # Mejora la tasa de acierto (Test-Time Augmentation)
)

# 3. Localizar el video guardado
output_dir = results[0].save_dir

# Buscar el archivo generado automáticamente (sin importar si es .avi o .mp4)
archivos_salida = os.listdir(output_dir)
video_guardado = None

for archivo in archivos_salida:
    if archivo.endswith(('.mp4', '.avi', '.mov', '.mkv')):
        video_guardado = os.path.join(output_dir, archivo)
        break

# 4. Descargar el resultado
if video_guardado:
    print(f"Descargando video optimizado: {video_guardado}")
    files.download(video_guardado)
else:
    print("No se encontró el video generado.")
```
