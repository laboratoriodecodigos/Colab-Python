```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.linear_model import SGDClassifier
import imageio
import os
```

```python
X, y = make_blobs(n_samples=200, centers=2, random_state=42, cluster_std=2.5)
```

```python
model = SGDClassifier(loss='log_loss', max_iter=1, learning_rate='constant', eta0=0.1)

classes = np.unique(y)
```

```python
os.makedirs("frames", exist_ok=True)
```

```python
frames = []
n_steps = 30

for step in range(n_steps):

    # Entrenar poco a poco
    model.partial_fit(X, y, classes=classes)

    # Crear malla para frontera de decisión
    xx, yy = np.meshgrid(
        np.linspace(X[:,0].min()-1, X[:,0].max()+1, 100),
        np.linspace(X[:,1].min()-1, X[:,1].max()+1, 100)
    )

    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    # Graficar
    plt.figure(figsize=(5,5))
    plt.contourf(xx, yy, Z, alpha=0.3)
    plt.scatter(X[:,0], X[:,1], c=y)
    plt.title(f"Paso {step+1}: El modelo está aprendiendo...")

    filename = f"frames/frame_{step}.png"
    plt.savefig(filename)
    plt.close()

    frames.append(imageio.imread(filename))
```

```python
gif_path = "clasificador.gif"
imageio.mimsave(gif_path, frames, duration=0.3)

gif_path
```
