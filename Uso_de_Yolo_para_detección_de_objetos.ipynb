```python
!pip install -q ultralytics
import cv2
import time
from ultralytics import YOLO

# Cargar modelo YOLO
model = YOLO("yolov8n.pt")

# Abrir video
cap = cv2.VideoCapture("trafico.mp4")

# Propiedades del video
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps    = cap.get(cv2.CAP_PROP_FPS)

# Escritor de video
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter("video_yolo.mp4", fourcc, fps, (width, height))

prev_time = time.time()

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Inferencia YOLO
    results = model(frame)
    frame = results[0].plot()

    # FPS
    curr_time = time.time()
    fps_calc = 1 / (curr_time - prev_time)
    prev_time = curr_time

    cv2.putText(
        frame,
        f"FPS: {int(fps_calc)}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 255, 0),
        2
    )

    out.write(frame)

cap.release()
out.release()

print("✅ video_yolo.mp4 generado correctamente")


from google.colab import files
files.download("video_yolo.mp4")
```
